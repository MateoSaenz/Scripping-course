```python
# Create table

import sqlite3

conn = sqlite3.connect('db.sqlite') 

cursor = conn.cursor()

cursor.execute('''
  CREATE TABLE users (
    id INTEGER PRIMARY KEY,
    name TEXT, 
    email TEXT
  )
''')

conn.commit()
```

```python
# Insert data 

cursor.execute('''
  INSERT INTO users (name, email) 
  VALUES (?, ?)
''', ('John Doe', 'john@email.com'))

conn.commit()
```

```python
cursor.execute('''
  SELECT * FROM users
''')

for row in cursor:
  print(row)
```

```python
from faker import Faker 
faker = Faker()
names = [faker.name().split() for i in range (100)]
names = [name for name in names if len(name)== 2]
names
```

```python
# Insert in the Database

import sqlite3


connection = sqlite3.connect('sample.db')

table= 'CREATE TABLE people (id integer primary key, name TEXT, surname TEXT)'

cursor = connection.cursor()
cursor.execute(table)
connection.commit()
```

```python
insert_query = 'INSERT INTO people(name,surname) VALUES(?,?)'
cursor = connection.cursor()
for name in names:
    cursor.execute(insert_query,name)
connection.commit()
```

```python
select_query = 'SELECT * FROM people LIMIT 10'
for i in cursor.execute(select_query):
    print(i)
```

```python
# Process data in Python

results = cursor.execute('SELECT * FROM people')

for row in results:
  first_name = row[0] 
  last_name = row[1]
  print(f"{first_name} {last_name}")
```
